```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    'Variable':       ['Climate','Housing','Health','Crime','Transportation',
                       'Education','Arts','Recreation','Economy'],
    'PC1':            [0.190, 0.544, 0.782, 0.365, 0.585, 0.394, 0.985, 0.520, 0.142],
    'PC2':            [0.017, 0.020,-0.605, 0.294, 0.085,-0.273, 0.126, 0.402, 0.150],
    'PC3':            [0.207, 0.204, 0.144, 0.585, 0.234, 0.027,-0.111, 0.519, 0.239]
}
df = pd.DataFrame(data)
print(df)
```

```python
df['PC1_abs'] = abs(df['PC1'])
df['PC2_abs'] = abs(df['PC2'])
df['PC3_abs'] = abs(df['PC3'])

print("\nHighest loading in PC1:", df.loc[df['PC1_abs'].idxmax(), 'Variable'], "=", df['PC1_abs'].max())
print("Highest loading in PC2:", df.loc[df['PC2_abs'].idxmax(), 'Variable'], "=", df['PC2_abs'].max())
print("Highest loading in PC3:", df.loc[df['PC3_abs'].idxmax(), 'Variable'], "=", df['PC3_abs'].max())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for idx, (col, ax) in enumerate(zip(['PC1','PC2','PC3'], axes)):
    ax.barh(df['Variable'], abs(df[col]), color='steelblue')
    ax.set_title(f'{col} - Variable Loadings')
    ax.set_xlabel('Absolute Loading')
plt.tight_layout()
plt.show()
```

```python
print("Principal Component Analysis Results:")
print("=" * 45)
print("PC1 - Primary Principal Component:")
print("  → Arts (0.985) has the HIGHEST loading")
print("  → Arts is the most important variable in PC1")
print()
print("PC2 - Secondary Principal Component:")
print("  → Health (-0.605) has the highest absolute loading")
print("  → Health is most important in PC2")
print()
print("PC3 - Third Principal Component:")
print("  → Crime (0.585) has the highest loading")
print("  → Crime is most important in PC3")
print()
print("CONCLUSION: Arts is the PRINCIPAL COMPONENT (highest overall loading 0.985 in PC1)")
```
